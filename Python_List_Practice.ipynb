# Method in List

## append() method

```python
n1= [10, 20]
n1.append(30)
print(n1)
```

```python
n1= [20, 30, 40, 50, 60]
n1.append(101)
print(n1)
```

## extend() method

```python
n1= [1, 2, 3, 4, 5]
n1.extend([10,30])
print(n1)
```

```python
n1=["apple", "banana", "pineapple"]
n1.extend(["Orange", "blueberry"])
print(n1)
```

## insert() method

```python
lst=[2, 4, 6, 8, 10]
lst.insert(2,12)
print(lst)
```

```python
lst=["date", "time", "value"]
lst.insert(2,"calculaltion")
print(lst)
```

## remove() method

```python
n1=[4, 5, 6, 7, 8, 9]
n1.remove(4)
print(n1)
```

```python
s1=[0, 5, 7, 3, 7]
s1.remove(7)
print(s1)
```

# pop() method

```python
k1=[2, 4, 6, 8, 10]
k1.pop(3)
print(k1)
```

## clear() method

## sort() method
