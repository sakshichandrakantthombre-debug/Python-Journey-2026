```python
# WAP to hallow square using pattern logic:

n=4

for i in range(n):
    for j in range(n):
         if i == 0 or i== n-1 or j== 0 or j== n-1:
                print("*", end=" ")
         else:
             print(" ", end=" ")
    print()
```

```python
# wap pyrimid patterns with Alphabats:
print("GFG")
n = 5
alph = 65
for i in range(0, n):
    print(" " * (n-i), end=" ")
    for j in range(0, i+1):
        print(chr(alph), end=" ")
        alph += 1
        alph = 65
    print()
```

```python
# wap pyrimid patterns with numbers:
print("Number")
n = 5
for i in range(0, n):
    print(" " * (n-i), end=" ")
    for j in range(0, i+1):
        print(n, end=" ")
    print()
```

```python
# WAP to hallow  triangle using pattern logic

rows = int(input("Enter the row size for the pattern: "))
for i in range(1, rows + 1):  # Outer loop for rows
    for j in range(1, i + 1):  # Inner loop for columns
        if j == 1 or i == rows or i == j:  # Print star on borders
            print("*", end=" ")
        else:
            print(" ", end=" ")  # Print space inside
    print()
```

```python

#Inverted Number Right-Angled Triangle Pattern in Python:

rows=7
for i in range(rows, 0, -1):  # Outer loop for rows in reverse
    for j in range(1, i + 1):  # Inner loop for columns
        print(j, end=" ")  # Print numbers
    print()
```

```python
# wap cross pattern python:
rows =6
for i in range(1, rows + 1):  # Outer loop for rows
    for j in range(1, rows + 1):  # Inner loop for columns
        if i == j or i + j == rows + 1:  # Print stars in cross pattern
            print("*", end=" ")
        else:
            print(" ", end=" ")  # Print space
    print()
```
